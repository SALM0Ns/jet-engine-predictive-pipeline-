```python
import duckdb
```

```python
# 1. ชี้เป้าไปที่ไฟล์ 1 ล้านแถวที่เราเพิ่งสร้าง
PARQUET_FILE = "/Users/salmonm/Documents/Data Engineer Project/Airceaft Data/jet-engine-predictive-pipeline-/downloaded_sensor.parquet"

# 2. ลองเขียน SQL ถามว่า "เครื่องยนต์ไหน ร้อนที่สุดจากประวัติ 1 ล้านแถวนี้?"
query = f"""
    SELECT 
        *
    FROM '{PARQUET_FILE}'
    LIMIT 10
""" 

# 3. สั่งรันและปรินต์ออกมาดู!
print("Query 1 million row data...")
result = duckdb.query(query).df() # แปลงผลลัพธ์เป็น Pandas DataFrame
print(result)
```
